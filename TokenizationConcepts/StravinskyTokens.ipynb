```python
txt = "Pulcinella is a 21-section ballet by Igor Stravinsky with arias for soprano, tenor and bass vocal soloists, and two sung trios. It is based on the 18th-century play Quatre Polichinelles semblables, or Four similar Pulcinellas, revolving around a stock character from commedia dell'arte. The work premiered at the Paris Opera on 15 May 1920 under the baton of Ernest Ansermet. The central dancer, Léonide Massine, created both the libretto and the choreography, while Pablo Picasso designed the costumes and sets. The ballet was commissioned by Sergei Diaghilev, impresario of the Ballets Russes. A complete performance takes 35–40 minutes. Stravinsky revised the score in 1965."
```

```python
gpt2Tokenizer = AutoTokenizer.from_pretrained('gpt2');
```

```python
words = txt.split();

encoded_tokens = gpt2Tokenizer.encode(txt);

print(f"Words count = {len(words)}");
print(f"Tokens count = {len(encoded_tokens)}");
print(f"Unique Tokens = {len(set(encoded_tokens))}");



```

```python
decoded_token = gpt2Tokenizer.decode(encoded_tokens);
print(txt + ": test\n");
print(decoded_token + ": Decoded form tokens");

print(f"decoded txt = word?: {decoded_token == txt}");
print(encoded_tokens)

print(gpt2Tokenizer.decode(encoded_tokens[1]))
```

```python
title = ["Token", "Length"]
# data = [encoded_tokens, list(len(gpt2Tokenizer.decode(i)) for i in encoded_tokens)]
data = []

for i in encoded_tokens:
    data.append([i, len(gpt2Tokenizer.decode(i))])

pandasData = pd.DataFrame(data = data, columns=title);

print(pandasData.head())

bar_data = pandasData.value_counts(title[1]);

print(bar_data)
```

```python
# sns.set_theme();
sns.barplot(x=bar_data.index, y=bar_data.values, palette="plasma", legend=False, hue=bar_data.values);
# sns.axisgrid();
sns.despine();
plt.gca().set(xlabel='Token length (characters)',ylabel='Count');

```

```python

baseurl = 'https://www.gutenberg.org/cache/epub'

bookurls = [
    # code       title
    ['84',    'Frankenstein'    ],
    ['64317', 'GreatGatsby'     ],
    ['11',    'AliceWonderland' ],
    ['1513',  'RomeoJuliet'     ],
    ['76',    'HuckFinn'        ],
    ['219',   'HeartDarkness'   ],
    ['2591',  'GrimmsTales'     ],
    ['2148',  'EdgarAllenPoe'   ],
    ['36',    'WarOfTheWorlds'  ],
    ['829',   'GulliversTravels']
]
```

```python
bookContent = [];
for [code, title] in bookurls:
    txt = requests.get(f"{baseurl}/{code}/pg{code}.txt").text;
    bookContent.append(txt);
```

```python
bookContentToken = [];
title = ["Token", "Length"]

for txtContent in bookContent:
    # txtContent = bookContent[0];
    txtToken = gpt2Tokenizer.encode(txtContent);

    data = [];

    for charToken in txtToken:
        char = gpt2Tokenizer.decode(charToken);
        data.append([charToken, len(char)]);
    tokenDataFrame = pd.DataFrame(data = data, columns=title);
    bookContentToken.append(tokenDataFrame);


# for txtContent in bookContent:
#     txtToken = gpt2Tokenizer.encode(txtContent);
#     print(txtToken)
```

```python

print(len(bookContentToken));
print(bookContentToken[0].value_counts(title[1]))
```

```python

plt.figure(figsize=(10,4))
sns.set_theme(style="darkgrid")
markersymbols = 'soh^vP*DXp'

for i in range(len(bookContentToken)):
    # print(plotdata.value_counts(title[1]));
    sns.scatterplot(data=bookContentToken[i].value_counts(title[1], normalize=True), 
                    s = 80, 
                    label=bookurls[i][1],
                    markers=markersymbols[i],
                    edgecolor='k');

plt.gca().set(xlabel='Token length (characters)',ylabel='Frequency (norm.)',xscale='log');
plt.tight_layout()
plt.show();
```
